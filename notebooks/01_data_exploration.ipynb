```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate the project folder
PROJECT_DIR = Path.cwd().parent

# Locate the raw data folder
raw_folder = PROJECT_DIR / "data" / "raw"

# Find CSV files
csv_files = list(raw_folder.glob("*.csv"))

print("Project folder:", PROJECT_DIR)
print("CSV files found:", len(csv_files))

# Stop if the CSV is missing
if not csv_files:
    raise FileNotFoundError(
        "No CSV found. Check that the file is inside data/raw."
    )

# Load the dataset
file_path = csv_files[0]
df = pd.read_csv(file_path, encoding="latin1")

print("\nDataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
print("Total number of columns:", len(df.columns))
print("\nColumn names:\n")

for column in df.columns:
    print(column)
```

```python
# Count missing values in every column
missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)
```

```python
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())
```

```python
# Parse dates without modifying the original DataFrame
order_dates = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

ship_dates = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

print("Invalid order dates:", order_dates.isna().sum())
print("Invalid ship dates:", ship_dates.isna().sum())

print(
    "Orders shipped before order date:",
    (ship_dates < order_dates).sum()
)

print("\nNegative sales rows:", (df["Sales"] < 0).sum())
print("Negative quantity rows:", (df["Quantity"] < 0).sum())
print("Negative profit rows:", (df["Profit"] < 0).sum())
print(
    "Discounts outside 0–1:",
    (~df["Discount"].between(0, 1)).sum()
)
```

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Cleaning dataset shape:", df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[\s-]+", "_", regex=True)
)

print(df_clean.columns.tolist())
```

```python
text_columns = df_clean.select_dtypes(
    include=["str", "object", "string"]
).columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Whitespace cleanup completed.")
print("Text columns cleaned:", len(text_columns))
```

```python
df_clean["order_date"] = pd.to_datetime(
    df_clean["order_date"],
    errors="raise"
)

df_clean["ship_date"] = pd.to_datetime(
    df_clean["ship_date"],
    errors="raise"
)

print(df_clean[["order_date", "ship_date"]].dtypes)
print("\nFirst five order dates:")
print(df_clean["order_date"].head())
```

```python
print("Dataset shape:", df_clean.shape)

print("\nMissing values:", df_clean.isnull().sum().sum())

print("Exact duplicate rows:", df_clean.duplicated().sum())

print(
    "Duplicate row IDs:",
    df_clean["row_id"].duplicated().sum()
)

print(
    "Invalid order dates:",
    df_clean["order_date"].isna().sum()
)

print(
    "Invalid ship dates:",
    df_clean["ship_date"].isna().sum()
)

print("\nCleaned data types:")
df_clean.info()
```

```python
output_folder = PROJECT_DIR / "data" / "processed"
output_folder.mkdir(parents=True, exist_ok=True)

output_file = output_folder / "superstore_cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("Saved location:", output_file)
print("Rows exported:", len(df_clean))
```
